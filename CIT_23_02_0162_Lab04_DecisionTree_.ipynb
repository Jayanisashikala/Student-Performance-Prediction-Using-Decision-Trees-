# 🔵 **Introduction**
This mini project focuses on predicting student academic performance using a Decision Tree classifier. The goal is to identify whether a student will Pass or Fail based on academic and behavioral factors. Such a system helps educational institutions identify at risk students early and provide academic support.



# 🔵 **Dataset Description**
The dataset contains student academic and study related attributes.

Features:

*   **Attendance_Percentage** - Student attendance in percentage
*   **Midterm_Marks** - Marks obtained in midterm exam
*   **Assignment_Score** - Assignment marks
*   **Study_Hours_Per_Week** - Average weekly study hours
*   **Previous_GPA** - GPA from previous semester

Target Variable:

*   Pass
*   Fail




# 🔵 **Process of the Lab Task**

## Srep 1 - Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix
```

## Step 2 - Load the Dataset

```python
# Read the CSV file
df = pd.read_csv("Student_Performance.csv")
df
```

## Step 3 - Dataset Exploration

The dataset was loaded into a pandas DataFrame and basic exploration was performed.

*   Checked dataset shape
*   Checked for missing values
*   Verified data types and dataset information

This step helped to understand the structure and quality of the dataset.

```python
# Dataset shape
print("\nDataset shape:")
print(df.shape)

# Missing value
print("\nMissing values:")
print(df.isnull().sum())

# Dataset Information
print("\nDataset information:")
print(df.info())
```

## Step 4 - Basic Data Analysis

Basic statistical analysis was performed to understand the distribution of values.


*   Count,mean, standard deviation, minimum, and maximum values were examined
*   This analysis showed how attendance, marks, and GPA affect student performance


```python
# Statistical summary
df.describe()
```

## Step 5 - Encode the Target Variable

The target variable Final_Result was categorical (Pass / Fail).

It was encoded into numerical form so that the Decision Tree model could process it.
*   Pass - 1
*   Fail - 0





```python
# Encode target variable
df['Final_Result'] = df['Final_Result'].map({'Pass': 1, 'Fail': 0})

df.head()
```

## Step 6 - Define Features and Target

**X** contains the input features (Attendance_Percentage, Midterm_Marks, Assignment_Score, Study_Hours_Per_Week, and Previous_GPA) that are used to predict student performance.

**y** contains the target variable (Final_Result), which represents whether the student passed or failed.



```python
x = df[["Attendance_Percentage", "Midterm_Marks", "Assignment_Score", "Study_Hours_Per_Week", "Previous_GPA"]]
y = df["Final_Result"]
```

## Step 7 - Train-Test Split

The dataset was split into,

*   **Training set** – used to train the model
*   **Testing set** – used to evaluate performance

This ensures that the model is tested on unseen data.

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=42)
```

## Step 8 - Train the Decision Tree Classifier

A Decision Tree Classifier was used to train the model.

*   The model learned decision rules from the training data
*   Features such as attendance, marks, and GPA were used to predict the final result





```python
dt_model = DecisionTreeClassifier(max_depth=4, random_state=42)
```

```python
dt_model.fit(x_train, y_train)
```

## Step 9 - Make Predictions


*   y_pred stores the predicted results generated by the trained Decision Tree model using the test data.
*   The comparison DataFrame displays the actual values and predicted values side by side to easily compare model performance.



```python
y_pred = dt_model.predict(x_test)
```

```python
comparison = pd.DataFrame({"Actual": y_test.values, "Predicted": y_pred})
```

```python
comparison.head()
```

## Step 10 - Evaluate the Model

The trained model was evaluated using accuracy.
*   Accuracy measures how correctly the model predicts Pass or Fail
*   A good accuracy value indicates that the model performs well on test data



```python
accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)
```

```python
print("Model Accuracy:", accuracy)
print("Confusion Matrix:\n", cm)
```

## Step 11 - Overfitting Control

Overfitting occurs when the tree becomes too deep and learns noise from training data.

Underfitting occurs when the tree is too shallow and fails to capture important patterns.

Tree complexity can be controlled using parameters such as,

*   max_depth
*   min_samples_split
*   min_samples_leaf

```python
# Decision Tree with controlled depth
dt_limited = DecisionTreeClassifier(
    max_depth=3,
    min_samples_split=2,
    random_state=42
)

dt_limited.fit(x_train, y_train)

y_pred_limited = dt_limited.predict(x_test)

accuracy_limited = accuracy_score(y_test, y_pred_limited)

print("Accuracy with controlled tree:", accuracy_limited)
```
